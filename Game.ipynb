```python

print('*'*5,"Welcome To Number Guessing Game",'*'*5)
s_num = 74
atm = 1
while atm < 8:
    user_num = int(input("Enter your Guess :"))
    if user_num > 0 and user_num <= 100:
        if user_num > s_num:
            print(f"your Guess is Higher Then The Secret Number, you Have Remaining {8-atm} Attempts")
        elif user_num > s_num:
            print(f"your Guess is Lower Then The Secret Number, you Have Remaining {8-atm} Attempts")
        else:
            print(f"Welldone, You Find The Secret Numer {s_num} in {atm} Attempts!!!")
    else:
        print("please enter number between  1 to 100 only")
    atm += 1
if user_num == s_num:
    print('*'*5,"You Won The Game",'*'*5)
else:
    print("Your Attempts Are over Try Again  Later !!!")
```

```python
print('*'*5,"Welcome To Letter Guessing Game",'*'*5)
s_letter = "U"
atm = 1
while atm < 8:
    user_letter = input("Enter your Guess :").upper()
    if user_letter > "A" and user_letter <= "Z":
        if user_letter > s_letter:
            print(f"your Guess is Higher Then The Secret Letter, you Have Remaining {8-atm} Attempts")
        elif user_letter > s_letter:
            print(f"your Guess is Lower Then The Secret Letter, you Have Remaining {8-atm} Attempts")
        else:
            print(f"Welldone, You Find The Secret Numer {s_letter} in {atm} Attempts!!!")
            break
    else:
        print('please enter letter between "A" and "Z" only')
    atm += 1
if user_letter == s_letter:
    print('*'*5,"You Won The Game",'*'*5)
else:
    print("Your Attempts Are over Try Again  Later !!!")
```
