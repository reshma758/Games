For Loop
*A for loop in python is used to iterate over a sequence (such as list, tuple dictionary, set, or string) and exceute a block of code multiple times.

```python
"""
for variable in sequence:
    # block of code
    
```

```python
for num in range(1,21):
        print(num)
```

```python
# ODD num without condition
for num in range(1,21,2):
        print(num)
```

```python
# Even num without condition
for num in range(2,23,2):
        print(num)
```

```python
# ODD Even num with condition:
num = int(input("Enter your Range:"))
for i in range (1,num+1):
    if i % 2 == 0:
        print("Even number:",i)
```

```python
#Even num with condition:
num = int(input("Enter your Range:"))
print("Even number :")
for i in range (1,num+1):
    if i % 2 == 0:
        print(i)
```

```python
# ODD num with condition:
num = int(input("Enter your Range:"))
print("odd number :")
for i in range (1,num+1):
    if i % 2 != 0:
        print(i)
    
```

```python
# ODD Even num with condition:
num = int(input("Enter your Range:"))
for i in range (1,num+1):
    if i % 2 == 0:
        print("Even number:",i)
    else:
        print("Odd number:",i)
```

```python
# ODD Even num with condition:
num = int(input("Enter your Range:"))
odd =[]
even = []
for i in range (1,num+1):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even number:",even)
print("odd number:",odd)
    
```

```python
## for loop in string
text ="Hello this is Python"
for i in text:
    print(i,end = " ")
```

```python
text = "Hello this is Python"
text[::-1]
```

```python
text = "Hello this is Python"
count = 0
for i in text :
    count += 1
print(count)
```

```python
table = int(input("Enter your table:"))
for i in range(1,21):
    print(f'{i} X {table} = {i*table}')
```

```python
#listed
item = {'apple','banana'}
for i in item:
    print(i)
```

```python
# nested 
item = {'apple','banana'}
for i in item:
    for j in i:
        print(j,end="")
    print()
```
