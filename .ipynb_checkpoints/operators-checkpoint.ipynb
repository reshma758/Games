```python
num_1 = int(input("Enter Number 1 :"))
num_2 = int(input("Enter Number 2 :"))
print("Addition :",num_1 + num_2)
print("sub :",num_1 - num_2)
print("multiply :",num_1 * num_2)
print("division :",num_1 / num_2)
print("floor division :",num_1 // num_2)
print("modulas :",num_1 % num_2)
print("power :",num_1 ** num_2)
```


Comparison Operators

'=='

'>='

```python
num_1 = float(input("Enter Number 1 :"))
num_2 = float(input("Enter Number 2 :"))
print(num_1 > num_2)
print(num_1 < num_2)
print(num_1 >= num_2)
print(num_1 <= num_2)
print(num_1 == num_2)
print(num_1 != num_2)
```

### Bitwise operators
* '%'
* '|'
* '~'
* '^'
* '>>'
* '<<'
  

```python
a = 10 
b = 6
# 32 16 8 4 2 1 
#       1 0 1 0 --10
#       0 1 1 0 -- 6
print ( a & b) # and 2 value true  
print ( a | b) # or 1 value ans 14
print (~b) # not for a value -(n+1)
print (a ^ b) # XOR  1 1 0 0
print (a >> b)
print (a << b)
```

```python
num = 20
num += 5 # num = num + 5 = 20 + 5 = 25
num -= 2 # num = num - 2 = 25 - 2 = 23
num *= 3 # num = num * 3 = 23 * 3 = 69
num /= 2 # num = num / 2 = 69 / 2 = 34.5 (float divison)
num //= 2 # num = num // 2 = 34.5 // 2 = 17 (floor division)
num %= 3 # num = num % 3 = 17 % 3 = 2 (modulus)
num **= 3 # num = num ** 3 = 2 ** 3 = 8 (power)
```

```python
num
```

Membership operators
* in
* not in

```python
stg = " Hello this is Python"
word = "Hello this is python"
word1 = "this"
print("hello" in stg )
print("python" in word )
print("python" in stg )
print("This" in word1 )
print("is" in word)
print("Is" in word1)
print("h" not in stg)
print("z" not in word)
```

```python
a = [ "Reshu",23,20,"Lavani"]
print('Reshu' in a)
print('lavani' in a)
print ('23' in a)
print (23 in a)
print(22 not in a)
print(23 not in a)
```
