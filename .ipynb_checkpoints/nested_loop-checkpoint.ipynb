```python
row = int(input("Enter the number of rows: "))
for i in range(1, row +1): # i = 1
    for j in range(1, row + 1): # (i=1 >> j = 1 to 6 ) ( i = 2 >> j = 1 to 6)
        print("*", end=" ")
    print() # Move to the next line after each row
```

```python
row = int(input("Enter the number of rows: "))
for i in range(1, row +1):
    for j in range(1, row + 1):
        print(j, end=" ")
    print()
```
