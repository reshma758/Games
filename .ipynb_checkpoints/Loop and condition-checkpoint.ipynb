1) Even Numbers Printer Write a for loop to print all even numbers between 1 and 100.

```python
for i in range(1, 101):
    if i % 2 == 0:
        print(i)
```

2) Sum of Natural Numbers Use a while loop to calculate the sum of the first 50 natural numbers.

```python
num = 1
total = 0
while num <= 50:
    total = total + num
    num = num + 1 
print("Sum =", total)
```

3) Factorial Calculation Write a program using a loop to calculate the factorial of a given number n.

```python
n = int(input("Enter a number: "))
factorial = 1
for i in range(1, n + 1):
    factorial = factorial * i
print("Factorial =", factorial)
```

4) Multiplication Table Generate the multiplication table of 9 using a for loop.

```python
for i in range(1, 11):
    print(9 * i)
```

5) Reverse String Use a while loop to reverse the string "assignment" without using slicing.

```python
text = "Data analyst"
reverse = ""
i = len(text) - 1
while i >= 0:
    reverse = reverse + text[i]
    i = i - 1
print("Reversed string:", reverse)
```

6) Count Vowels Write a loop with conditional statements to count the number of vowels in the string "programming language".

```python
text = "programming language"
vowels = "aeiou"
count = 0
for char in text:
    if char in vowels:
        count = count + 1
print("Number of vowels:", count)
```

7) Fibonacci sequence is a series where each number is the sum of the two preceding ones: 0, 1, 1, 2, 3, 5, 8, 13, …

```python
n = int(input("Enter number of terms: "))
a = 0
b = 1
c = 2
for i in range(n):
    print(a)
    c = a + b
    a = b
    b = c
```

8) Palindrome Concept
* A palindrome is a word, phrase, or number that reads the same backward as forward. Examples: "madam", "racecar", 121.


```python
text = input("Enter a word: ")
reverse = ""
for char in text:
    reverse = char + reverse
if text == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

9. Armstrong Number


```python
num = int(input("Enter a number: "))
original = num
total = 0
while num > 0:
    digit = num % 10
    total = total + digit ** 3
    num = num // 10
if total == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

10. Print all prime numbers between 1 and 100.



```python
for num in range(2, 101):
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1
    if count == 2:
        print(num)
```
