### Error or Exception Handling
* try
* except
* else
* finally

```python
def div(a,b):
    return a / b
```

```python
div(5,2)
```

```python
try:
    def div(a,b):
        return a / b
except ZeroDivisionError:
    print("This is ZeroDivisionError:")
```

```python
div(5,0)
```

```python
def div(a,b):
    try:
        return a / b
    except ZeroDivisionError:
        return 'ZeroDivisionError:'
```

```python
div(5,0)
```

```python
def div(a,b):
    try:
        return a / b
    except ZeroDivisionError:
        return 'ZeroDivisionError:'
    except TypeError:
        return 'TypeError:'
```

```python
div("5",0)
```

```python
def divt(a,b):
    try:
        return a / b
    except Exception as e:
        print("Unexcepted error: e")
        return[]
```

```python
divt(5,0)
```
