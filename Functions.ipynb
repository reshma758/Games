```python
# positional args
def pos_arg(*args):
    for i in args:
        print(i)
```

```python
a = [10,20,30,40,50,"Jp"]
```

```python
pos_arg(10,20,30,40,50,"Jp")
```

```python
# keyword args
def key_args(**kwargs):
    for k,v in kwargs.items():
        print(f'{k} : {v}')
```

```python
key_args(name = "Reshma",age = 24,city = "chennai")
```

```python
# postional with keyword arg
def key_pos_arg(*args,**kwargs):
    for i in args:
        print(i)
    print("Keyword arg")
    for k , v in kwargs.items():
        print(f"{k} : {v}")
```

```python
key_pos_arg(10,20,30,"Reshu", name = "Reshma", age = 24, city = "chennai")
```
