Conditional statement
* if -- one condition one way exceution
* if else -- one condition two way exceution
* if elif else -- Multiple conditon multiple way of exceution
* nested if

```python
### Conditional statement 
"""
if condition:
    # statement
"""
age = int(input("Enter your age :"))
if age >= 18:
        print("Eligible")
```

```python
"""
if condition:
    # statement
else:
    # statement
"""
age = int(input("Enter your age :"))
if age >= 18:
    print("Eligible for vote")
else:
    print("Not Eligible for vote")
```

```python
"""
if condition:
    # statement
else:
    # statement
elise:
    # statement
"""
total_mark = float(input("Enter your mark % :"))
if total_mark >= 90:
    print("S Grade")
elif total_mark >= 80:
    print("A Grade")
elif total_mark >= 70:
    print("B Grade")
elif total_mark >= 60:
    print("c Grade")
elif total_mark >= 50:
    print("D Grade")
else:
    print("E Grade")
    
```
